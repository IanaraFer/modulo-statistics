## Problem 1: Extending the Lady Tasting Tea

Let's extend the Lady Tasting Tea experiment as follows. The original experiment has 8 cups: 4 tea-first and 4 milk-first. Suppose we prepare 12 cups: 8 tea-first and 4 milk-first. A participant claims they can tell which was poured first.

We will:
1. Simulate this experiment using numpy by randomly shuffling the cups many times
2. Calculate the probability of the participant correctly identifying all cups by chance
3. Compare our result with the original 8-cup experiment
4. Discuss whether to extend or relax the p-value threshold compared to the original design


```python
# Import necessary libraries
import numpy as np
from scipy.special import comb
```

### Understanding the Experiment

In the original Lady Tasting Tea experiment:
- **8 cups total**: 4 tea-first, 4 milk-first
- The participant must correctly identify all 8 cups
- The probability of getting all correct by chance is 1 / C(8,4) = 1/70 ≈ 0.0143 (1.43%)

In our extended experiment:
- **12 cups total**: 8 tea-first, 4 milk-first
- The participant must correctly identify all 12 cups
- The probability of getting all correct by chance is 1 / C(12,8) = 1/495 ≈ 0.00202 (0.202%)

We'll verify this through simulation.


```python
# Set up the experiment parameters
# Extended experiment: 12 cups (8 tea-first, 4 milk-first)
n_cups_extended = 12
n_tea_first_extended = 8
n_milk_first_extended = 4

# Original experiment: 8 cups (4 tea-first, 4 milk-first)
n_cups_original = 8
n_tea_first_original = 4
n_milk_first_original = 4

# Number of simulations to run
n_simulations = 1000000
```

### Simulation Process

We'll simulate the experiment by:
1. Identifying the actual positions of tea-first cups (positions 0-7 in our setup)
2. For each simulation, randomly selecting 8 positions out of 12 (representing the participant's random guess)
3. Checking if the randomly selected positions match the actual tea-first positions
4. Calculating the probability as the proportion of perfect matches

The key insight: If the participant is guessing randomly, they are essentially randomly choosing 8 positions out of 12 to identify as tea-first. The probability they get all correct is 1 / C(12,8) = 1/495, where C(12,8) is the number of ways to choose 8 positions out of 12.


```python
# Create the actual cup configuration for extended experiment
# 1 represents tea-first, 0 represents milk-first
actual_cups_extended = np.array([1] * n_tea_first_extended + [0] * n_milk_first_extended)

# Create the actual cup configuration for original experiment
actual_cups_original = np.array([1] * n_tea_first_original + [0] * n_milk_first_original)

print(f"Extended experiment configuration: {actual_cups_extended}")
print(f"Original experiment configuration: {actual_cups_original}")
```

```python
# Helper function to simulate perfect match probability
# Uses a vectorized approach over simulations for efficiency.
from typing import Tuple

def simulate_exact_match(n_cups: int, n_tea: int, n_simulations: int, seed: int = 42) -> Tuple[int, float]:
    """Simulate the probability of randomly guessing all tea-first cups correctly.

    Parameters
    ----------
    n_cups : int
        Total number of cups.
    n_tea : int
        Number of tea-first cups.
    n_simulations : int
        Number of random trials to simulate.
    seed : int, default 42
        Seed for reproducibility.

    Returns
    -------
    perfect_matches : int
        Count of simulations where the random guess was exactly correct.
    estimated_prob : float
        Estimated probability of a perfect match.
    """
    rng = np.random.default_rng(seed)
    actual_positions = set(range(n_tea))  # Assume tea-first occupy initial indices
    perfect_matches = 0
    for _ in range(n_simulations):
        guessed = set(rng.choice(n_cups, size=n_tea, replace=False))
        if guessed == actual_positions:
            perfect_matches += 1
    estimated_prob = perfect_matches / n_simulations
    return perfect_matches, estimated_prob
```

```python
# Run simulation for extended experiment using helper function
np.random.seed(42)  # For reproducibility of any legacy numpy calls
perfect_matches_extended, estimated_prob_extended = simulate_exact_match(
    n_cups=n_cups_extended,
    n_tea=n_tea_first_extended,
    n_simulations=n_simulations,
    seed=42
)

print(f"Perfect matches (extended): {perfect_matches_extended}")
print(f"Estimated probability (extended): {estimated_prob_extended:.6f} ({estimated_prob_extended*100:.4f}%)")
```

```python
# Run simulation for original experiment using helper function
perfect_matches_original, estimated_prob_original = simulate_exact_match(
    n_cups=n_cups_original,
    n_tea=n_tea_first_original,
    n_simulations=n_simulations,
    seed=123  # Different seed to ensure independence
)

print(f"Perfect matches (original): {perfect_matches_original}")
print(f"Estimated probability (original): {estimated_prob_original:.6f} ({estimated_prob_original*100:.4f}%)")
```

### Theoretical Probability Calculation

Let's also calculate the theoretical probabilities to compare with our simulation results.


```python
# Calculate theoretical probabilities using combinations
# The probability of correctly identifying all cups by chance is 1 / C(n, k)
# where n is total cups and k is the number of tea-first cups

# Extended experiment: C(12, 8) = C(12, 4) = 495
theoretical_prob_extended = 1 / comb(12, 8)

# Original experiment: C(8, 4) = 70
theoretical_prob_original = 1 / comb(8, 4)

print(f"Theoretical probability (extended): {theoretical_prob_extended:.6f} ({theoretical_prob_extended * 100:.4f}%)")
print(f"Theoretical probability (original): {theoretical_prob_original:.6f} ({theoretical_prob_original * 100:.4f}%)")
print(f"\nNumber of possible arrangements:")
print(f"Extended experiment: C(12, 8) = {comb(12, 8):.0f}")
print(f"Original experiment: C(8, 4) = {comb(8, 4):.0f}")
```

### Comparison and Results


```python
# Compare simulation results with theoretical values
print("=" * 60)
print("COMPARISON OF RESULTS")
print("=" * 60)
print(f"\nExtended Experiment (12 cups: 8 tea-first, 4 milk-first):")
print(f"  Theoretical probability: {theoretical_prob_extended:.6f} ({theoretical_prob_extended * 100:.4f}%)")
print(f"  Simulated probability:   {estimated_prob_extended:.6f} ({estimated_prob_extended * 100:.4f}%)")
print(f"  Difference:              {abs(theoretical_prob_extended - estimated_prob_extended):.6f}")
print(f"\nOriginal Experiment (8 cups: 4 tea-first, 4 milk-first):")
print(f"  Theoretical probability: {theoretical_prob_original:.6f} ({theoretical_prob_original * 100:.4f}%)")
print(f"  Simulated probability:   {estimated_prob_original:.6f} ({estimated_prob_original * 100:.4f}%)")
print(f"  Difference:              {abs(theoretical_prob_original - estimated_prob_original):.6f}")

ratio = theoretical_prob_extended / theoretical_prob_original
print(f"\nStringency ratio (extended / original): {ratio:.2f}x")
print(f"Chance of success is {1/ratio:.2f}x smaller in extended experiment.")
```

### Interpretation and Discussion

**Key Findings:**

1. **Extended Experiment Probability**: The probability of correctly identifying all 12 cups (8 tea-first, 4 milk-first) by chance is approximately **0.202%** (1 in 495), compared to **1.43%** (1 in 70) for the original 8-cup experiment.

2. **Comparison**: The extended experiment is **7.07 times more stringent** than the original. This means:
   - It's much harder to get all cups correct by chance
   - A participant who succeeds in the extended experiment provides stronger evidence of their ability
   - The p-value threshold becomes more conservative

3. **Simulation Accuracy**: Our simulation with 1,000,000 trials provides a good approximation of the theoretical probability, confirming our understanding of the combinatorial nature of the problem.

**Implications for p-value Threshold:**

The original 8-cup experiment uses a p-value threshold of approximately 0.0143 (1.43%). In the extended 12-cup experiment:

- **Option 1: Keep the same threshold (0.0143)**: This would be more lenient relative to the actual probability. A participant would need to get all cups correct, which has a probability of 0.00202, well below the threshold. This maintains consistency in the interpretation but is actually more conservative than intended.

- **Option 2: Use the actual probability (0.00202)**: This would maintain the same level of stringency as the original experiment. A participant must get all cups correct to demonstrate ability, with the same relative difficulty.

- **Option 3: Relax the threshold**: If we wanted to maintain similar absolute difficulty, we could relax the threshold, but this would go against the principle of hypothesis testing.

**Recommendation:**

I would recommend **keeping the requirement that the participant must correctly identify all cups**, which gives a p-value of approximately **0.00202 (0.202%)**. This is more stringent than the original, which is appropriate because:
- It provides stronger evidence when the participant succeeds
- It maintains the same experimental structure (all-or-nothing)
- The lower probability makes it harder to achieve by chance, strengthening the conclusion if the participant succeeds

However, if the goal is to maintain a similar p-value threshold (around 0.01-0.05), we might consider allowing the participant to get 11 out of 12 correct, which would have a higher probability. But this would change the nature of the experiment from the original design.
